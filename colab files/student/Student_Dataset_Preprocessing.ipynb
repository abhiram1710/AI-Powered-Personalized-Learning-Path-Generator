```python
# Remove the Course column

df.drop(columns=["Course"], inplace=True)

print("Course column removed successfully.")
```

```python
df.columns
```

```python
# Forward fill the Sr.No. column

df["Sr.No."] = df["Sr.No."].ffill()

print("Missing values in Sr.No. :", df["Sr.No."].isnull().sum())
```

```python
# Display the first 10 rows

df.head(10)
```

```python
# Remove newline characters and leading/trailing spaces

df["Job profession"] = (
    df["Job profession"]
    .str.replace("\n", "", regex=False)
    .str.strip()
)

print("Job profession column cleaned successfully.")
```

```python
# Display the first 10 job professions

df["Job profession"].head(10)
```

```python
# Convert s/p values to uppercase

df["s/p"] = df["s/p"].str.upper()

print("s/p column standardized successfully.")
```

```python
df[["Student", "s/p"]].head(10)
```

```python
# Rows with missing values in P6

df[df["P6"].isnull()]
```

```python
# Rows with missing values in P8

df[df["P8"].isnull()]
```

```python
# Fill missing values using the mode

df["P6"] = df["P6"].fillna(df["P6"].mode()[0])
df["P8"] = df["P8"].fillna(df["P8"].mode()[0])

print("Missing values handled successfully.")
```

```python
# Check remaining missing values

df.isnull().sum()
```

```python
# Rename columns for consistency

df.rename(columns={
    "Sr.No.": "Sr_No",
    "Job profession": "Job_Profession",
    "Logical - Mathematical": "Logical_Mathematical",
    "Spatial-Visualization": "Spatial_Visualization",
    "s/p": "Student_ID"
}, inplace=True)

print("Column names renamed successfully.")
```
