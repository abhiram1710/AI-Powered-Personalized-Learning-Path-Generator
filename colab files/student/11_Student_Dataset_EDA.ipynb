```python
# Count of each job profession

job_counts = df["Job_Profession"].value_counts()

job_counts.head(10)
```

```python
# Plot job profession distribution

plt.figure(figsize=(12, 18))

df["Job_Profession"].value_counts().plot(kind="barh")

plt.title("Distribution of Job Professions")
plt.xlabel("Number of Students")
plt.ylabel("Job Profession")

plt.tight_layout()
plt.show()
```

```python
# List of intelligence attributes

intelligence_columns = [
    "Linguistic",
    "Musical",
    "Bodily",
    "Logical_Mathematical",
    "Spatial_Visualization",
    "Interpersonal",
    "Intrapersonal",
    "Naturalist"
]

intelligence_columns
```

```python
# Plot histograms for all intelligence attributes

df[intelligence_columns].hist(
    figsize=(16, 12),
    bins=10,
    edgecolor="black"
)

plt.suptitle("Distribution of Multiple Intelligence Scores", fontsize=16)
plt.tight_layout()
plt.show()
```

```python
# Boxplots for intelligence attributes

plt.figure(figsize=(14, 8))

sns.boxplot(data=df[intelligence_columns])

plt.title("Boxplot of Multiple Intelligence Scores")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Correlation heatmap for intelligence attributes

plt.figure(figsize=(10, 8))

correlation_matrix = df[intelligence_columns].corr()

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap of Multiple Intelligence Scores")
plt.tight_layout()
plt.show()
```

```python
# Display the frequency of ratings for each performance indicator

performance_columns = ["P1", "P2", "P3", "P4", "P5", "P6", "P7", "P8"]

for col in performance_columns:
    print(f"\n{col}")
    print(df[col].value_counts())
```

```python
# Visualize performance indicator distribution

performance_counts = pd.DataFrame({
    col: df[col].value_counts()
    for col in performance_columns
}).T

performance_counts.plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Distribution of Performance Indicators (P1–P8)")
plt.xlabel("Performance Indicators")
plt.ylabel("Number of Students")
plt.xticks(rotation=0)
plt.legend(title="Performance")
plt.tight_layout()
plt.show()
```

```python
# Calculate average score for each intelligence attribute

average_scores = df[intelligence_columns].mean().sort_values(ascending=False)

average_scores
```

```python
# Plot average intelligence scores

average_scores.plot(
    kind="bar",
    figsize=(10, 6),
    edgecolor="black"
)

plt.title("Average Multiple Intelligence Scores")
plt.xlabel("Intelligence Type")
plt.ylabel("Average Score")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
