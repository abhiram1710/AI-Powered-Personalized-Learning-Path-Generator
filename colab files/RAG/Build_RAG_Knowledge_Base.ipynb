```python
import pandas as pd
import numpy as np
import faiss
import pickle

from sentence_transformers import SentenceTransformer
```

```python
df = pd.read_csv("/content/archive (3).zip")

df.head()
```

```python
documents = df["job_description"].tolist()

print("Total Documents:", len(documents))
print(documents[0])
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
embeddings = model.encode(
    documents,
    show_progress_bar=True,
    convert_to_numpy=True
)

print(embeddings.shape)
```

```python
dimension = embeddings.shape[1]

index = faiss.IndexFlatL2(dimension)

index.add(embeddings)

print("Total vectors stored:", index.ntotal)
```

```python
faiss.write_index(index, "knowledge_base.index")
```

```python
with open("knowledge_documents.pkl","wb") as f:
    pickle.dump(documents,f)
```

```python
query = "Python programming for machine learning"

query_embedding = model.encode([query], convert_to_numpy=True)

distance, index_ids = index.search(query_embedding, k=5)

print(index_ids)
```

```python
for i in index_ids[0]:
    print("="*80)
    print(documents[i])
```
