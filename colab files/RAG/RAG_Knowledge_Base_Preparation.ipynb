```python
employment_df["text"] = (
    "Course Name: " + employment_df["name"].astype(str) +
    "\nInstitution: " + employment_df["institution"].astype(str) +
    "\nRating: " + employment_df["average_rating"].astype(str) +
    "\nReviews: " + employment_df["review_count"].astype(str) +
    "\nKnowledge: " + employment_df["knowledge"].astype(str)
)

print(employment_df["text"].head(2).to_string(index=False))
```

```python
core_df["text"] = (
    "Category: " + core_df["category"].astype(str) +
    "\nJob Title: " + core_df["job_title"].astype(str) +
    "\nJob Description: " + core_df["job_description"].astype(str) +
    "\nSkills: " + core_df["job_skill_set"].astype(str)
)

print(core_df["text"].head(2).to_string(index=False))
```

```python
all_documents = pd.concat(
    [
        student_df[["text"]],
        career_df[["text"]],
        employment_df[["text"]],
        core_df[["text"]]
    ],
    ignore_index=True
)

print("Total documents:", len(all_documents))
print(all_documents.head(3))
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

chunks = text_splitter.create_documents(
    all_documents["text"].tolist()
)

print("Total chunks:", len(chunks))
print("\nFirst chunk:")
print(chunks[0].page_content)
```

```python
!pip install -q sentence-transformers
```

```python
from sentence_transformers import SentenceTransformer

embedding_model = SentenceTransformer("all-MiniLM-L6-v2")

print("Embedding model loaded successfully")
```

```python
from sentence_transformers import SentenceTransformer

embedding_model = SentenceTransformer(
    "all-MiniLM-L6-v2",
    device="cuda"
)

print("Embedding model loaded successfully")
```

```python
texts = [chunk.page_content for chunk in chunks]

embeddings = embedding_model.encode(
    texts,
    batch_size=128,
    show_progress_bar=True,
    device="cuda"
)

print("Embedding shape:", embeddings.shape)
```

```python
!pip install -q faiss-cpu
```

```python
import faiss
import numpy as np

embedding_matrix = np.asarray(embeddings).astype("float32")

index = faiss.IndexFlatL2(embedding_matrix.shape[1])
index.add(embedding_matrix)

print("FAISS index created successfully")
print("Number of vectors:", index.ntotal)
print("Vector dimension:", index.d)
```

```python
import os
import pickle

rag_path = "/content/drive/MyDrive/AI POWERED PERSONALISED LEARNING PATH GENERATOR/Dataset/Processed/RAG"

os.makedirs(rag_path, exist_ok=True)

# Save FAISS index
faiss.write_index(
    index,
    os.path.join(rag_path, "knowledge_base.index")
)

# Save chunks
with open(os.path.join(rag_path, "knowledge_chunks.pkl"), "wb") as f:
    pickle.dump(chunks, f)

print("RAG knowledge base saved successfully!")
print("Files saved:")
print("- knowledge_base.index")
print("- knowledge_chunks.pkl")
```
