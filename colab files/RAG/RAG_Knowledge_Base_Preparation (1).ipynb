```python
import os

print("Searching for the original RAG .txt files...\n")

search_locations = [
    "/content",
    "/content/drive/MyDrive"
]

target_files = {
    "cloud.txt",
    "data_science.txt",
    "deep_learning.txt",
    "generative_ai.txt",
    "machine_learning.txt",
    "python.txt",
    "web_development.txt"
}

found_files = []

for location in search_locations:
    for root, dirs, files in os.walk(location):
        for file in files:
            if file.lower() in target_files:
                path = os.path.join(root, file)
                found_files.append(path)

print("Found files:", len(found_files))

for path in found_files:
    print(path)
```

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

chunks = []

for document in all_documents:
    split_docs = text_splitter.create_documents([document["text"]])

    for doc in split_docs:
        chunks.append({
            "content": doc.page_content,
            "source": document["source"]
        })

print("Total chunks:", len(chunks))
print("\nFirst chunk:")
print(chunks[0]["content"])
```

```python
from sentence_transformers import SentenceTransformer
import numpy as np

# Load embedding model
embedding_model = SentenceTransformer("all-MiniLM-L6-v2")

# Extract text from the 41 chunks
knowledge_texts = [chunk["content"] for chunk in chunks]

# Generate embeddings
knowledge_embeddings = embedding_model.encode(
    knowledge_texts,
    convert_to_numpy=True
)

print("Number of embeddings:", knowledge_embeddings.shape[0])
print("Embedding dimension:", knowledge_embeddings.shape[1])
```

```python
import faiss
import numpy as np

# Convert embeddings to float32
embedding_array = np.asarray(
    knowledge_embeddings,
    dtype="float32"
)

# Create FAISS index
index = faiss.IndexFlatL2(
    embedding_array.shape[1]
)

# Add embeddings
index.add(embedding_array)

print("FAISS index created successfully!")
print("Total vectors:", index.ntotal)
print("Vector dimension:", index.d)
```

```python
import pickle
import faiss

# Save FAISS index
faiss.write_index(index, "/content/faiss_index.index")

# Save knowledge chunks
with open("/content/knowledge_chunks.pkl", "wb") as f:
    pickle.dump(chunks, f)

print("FAISS index saved successfully!")
print("Knowledge chunks saved successfully!")
print("Total chunks saved:", len(chunks))
```

```python
def ask_rag(question, k=3):

    # 1. Convert question into embedding
    query_embedding = embedding_model.encode([question])
    query_embedding = np.array(query_embedding).astype("float32")

    # 2. Search FAISS
    distances, indices = index.search(query_embedding, k)

    # 3. Create context from retrieved chunks
    context = "\n\n".join(
        [chunks[idx]["content"] for idx in indices[0]]
    )

    # 4. Create prompt
    prompt = f"""
You are an AI learning assistant.

Answer the student's question using ONLY the information
provided in the context below.

Context:
{context}

Student Question:
{question}

Give a clear, simple and useful answer suitable for a student.
"""

    # 5. Generate answer using Gemini
    response = client.models.generate_content(
        model="gemini-3.6-flash",
        contents=prompt
    )

    return response.text
```

```python
question = "What should I learn to become a data scientist?"

answer = ask_rag(question)

print("Question:", question)
print("\nRAG Answer:\n")
print(answer)
```
