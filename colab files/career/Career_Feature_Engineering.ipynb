```python
import pandas as pd
import ast
```

```python
df = pd.read_csv(
    "/content/drive/MyDrive/AI POWERED PERSONALISED LEARNING PATH GENERATOR/Dataset/Processed/career_dataset_cleaned.csv"
)
```

```python
df["job_skill_set"] = df["job_skill_set"].apply(ast.literal_eval)
```

```python
df.head()
```

```python
df["job_skill_set"].head()
```

```python
def normalize_skills(skill_list):
    normalized = []

    for skill in skill_list:
        skill = skill.strip()      # Remove leading/trailing spaces
        skill = skill.title()      # Standardize capitalization
        normalized.append(skill)

    return normalized

# Apply normalization
df["job_skill_set"] = df["job_skill_set"].apply(normalize_skills)
```

```python
df["job_skill_set"].head()
```

```python
from collections import Counter

all_skills = []

for skills in df["job_skill_set"]:
    all_skills.extend(skills)

print("Total Skills:", len(all_skills))
print("Unique Skills:", len(set(all_skills)))
```

```python
career_statistics = career_profiles.copy()

career_statistics["description_length"] = career_statistics["job_description"].apply(len)
```

```python
posting_count = (
    df.groupby("job_title")
      .size()
      .reset_index(name="posting_count")
)
```

```python
career_statistics = career_statistics.merge(
    posting_count,
    on="job_title",
    how="left"
)
```

```python
career_statistics = career_statistics[
    [
        "category",
        "job_title",
        "num_skills",
        "description_length",
        "posting_count"
    ]
]
```

```python
career_statistics.to_csv(
    "/content/drive/MyDrive/AI POWERED PERSONALISED LEARNING PATH GENERATOR/Dataset/Processed/career_statistics.csv",
    index=False
)

print("Career statistics saved successfully!")
```

```python
career_knowledge_base = career_profiles.merge(
    career_statistics[
        ["job_title", "description_length", "posting_count"]
    ],
    on="job_title",
    how="left"
)
```

```python
career_knowledge_base = career_knowledge_base.rename(
    columns={
        "job_skill_set": "skills"
    }
)
```
