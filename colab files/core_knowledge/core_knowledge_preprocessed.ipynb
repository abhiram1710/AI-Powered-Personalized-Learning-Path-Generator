```python
import pandas as pd
import numpy as np
import re
```

```python
df = pd.read_csv("/content/archive (3).zip")
```

```python
df.head()
```

```python
df.drop(columns=["job_id"], inplace=True)
```

```python
df.drop(columns=["job_id", "skill_id"], inplace=True, errors='ignore')
```

```python
df.shape
```

```python
df.info()
```

```python
df.columns
```

```python
df.isnull().sum()
```

```python
df.fillna("Unknown", inplace=True)
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(" ", "_")
```

```python
text_columns = df.select_dtypes(include='object').columns

for col in text_columns:
    df[col] = df[col].str.lower()
```

```python
for col in text_columns:
    df[col] = df[col].str.strip()
```

```python
for col in text_columns:
    df[col] = df[col].str.replace(r'[^a-zA-Z0-9, ]', '', regex=True)
```

```python
skill_mapping = {
    "ml": "machine learning",
    "ai": "artificial intelligence",
    "js": "javascript",
    "py": "python"
}

df.replace(skill_mapping, inplace=True)
```
