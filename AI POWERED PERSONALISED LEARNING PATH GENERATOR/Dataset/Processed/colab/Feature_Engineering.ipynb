```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("/content/archive (3).zip")
```

```python
df.head()
```

```python
df.columns
```

```python
def categorize(skill):
    skill = str(skill).lower()

    if "python" in skill or "java" in skill or "c++" in skill:
        return "Programming"

    elif "sql" in skill or "mysql" in skill:
        return "Database"

    elif "machine learning" in skill or "deep learning" in skill:
        return "Artificial Intelligence"

    elif "power bi" in skill or "tableau" in skill:
        return "Data Visualization"

    elif "aws" in skill or "azure" in skill:
        return "Cloud Computing"

    else:
        return "Other"
```

```python
df["domain"] = df["job_skill_set"].apply(categorize)
```

```python
def difficulty(skill):

    skill = str(skill).lower()

    if any(x in skill for x in ["html","css","excel","python"]):
        return "Beginner"

    elif any(x in skill for x in ["sql","power bi","tableau","javascript"]):
        return "Intermediate"

    elif any(x in skill for x in ["machine learning","deep learning","nlp","computer vision"]):
        return "Advanced"

    else:
        return "Intermediate"

df["difficulty_level"] = df["job_skill_set"].apply(difficulty)

df[["job_skill_set","difficulty_level"]].head()
```

```python
def prerequisite(skill):

    skill = str(skill).lower()

    if "machine learning" in skill:
        return "Python, Statistics"

    elif "deep learning" in skill:
        return "Machine Learning"

    elif "power bi" in skill:
        return "Excel"

    elif "tableau" in skill:
        return "SQL"

    elif "cloud" in skill:
        return "Networking"

    else:
        return "None"

df["prerequisite"] = df["job_skill_set"].apply(prerequisite)

df[["job_skill_set","prerequisite"]].head()
```

```python
df["keywords"] = (
    df["category"].astype(str) + ", " +
    df["job_skill_set"].astype(str) + ", " +
    df["domain"].astype(str)
)

df[["keywords"]].head()
```

```python
df["knowledge_document"] = (
    "Job Title: " + df["job_title"].astype(str) +
    ". Category: " + df["category"].astype(str) +
    ". Skills: " + df["job_skill_set"].astype(str) +
    ". Domain: " + df["domain"].astype(str) +
    ". Difficulty: " + df["difficulty_level"].astype(str) +
    ". Prerequisite: " + df["prerequisite"].astype(str) +
    ". Description: " + df["job_description"].astype(str)
)

df[["knowledge_document"]].head()
```
