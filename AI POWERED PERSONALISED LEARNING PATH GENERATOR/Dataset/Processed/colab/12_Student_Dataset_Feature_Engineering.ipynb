```python
# Create intelligence profiles

student_intelligence_profiles = df[[
    "Student_ID",
    "Linguistic",
    "Musical",
    "Bodily",
    "Logical_Mathematical",
    "Spatial_Visualization",
    "Interpersonal",
    "Intrapersonal",
    "Naturalist"
]].copy()

student_intelligence_profiles.head()
```

```python
# Intelligence columns

intelligence_columns = [
    "Linguistic",
    "Musical",
    "Bodily",
    "Logical_Mathematical",
    "Spatial_Visualization",
    "Interpersonal",
    "Intrapersonal",
    "Naturalist"
]

# Create Dominant Intelligence feature

df["Dominant_Intelligence"] = df[intelligence_columns].idxmax(axis=1)

df[["Student_ID", "Dominant_Intelligence"]].head(10)
```

```python
# Count of dominant intelligence types

df["Dominant_Intelligence"].value_counts()
```

```python
# Create Average Intelligence Score

df["Average_Intelligence_Score"] = (
    df[intelligence_columns]
    .mean(axis=1)
    .round(2)
)

df[[
    "Student_ID",
    "Average_Intelligence_Score"
]].head(10)
```

```python
# Summary statistics

df["Average_Intelligence_Score"].describe()
```

```python
# Convert performance ratings into numerical scores

performance_mapping = {
    "POOR": 1,
    "AVG": 2,
    "BEST": 3
}

performance_columns = [
    "P1", "P2", "P3", "P4",
    "P5", "P6", "P7", "P8"
]

performance_scores = df[performance_columns].replace(performance_mapping)

# Create Average Performance feature

df["Average_Performance"] = (
    performance_scores
    .mean(axis=1)
    .round(2)
)

df[[
    "Student_ID",
    "Average_Performance"
]].head(10)
```

```python
# Summary statistics

df["Average_Performance"].describe()
```

```python
# Create Performance Category

def performance_category(score):
    if score >= 2.30:
        return "Excellent"
    elif score >= 2.00:
        return "Good"
    else:
        return "Average"

df["Performance_Category"] = df["Average_Performance"].apply(performance_category)

df[[
    "Student_ID",
    "Average_Performance",
    "Performance_Category"
]].head(10)
```

```python
# Count of performance categories

df["Performance_Category"].value_counts()
```

```python
# Create student statistics

student_statistics = pd.DataFrame({
    "Metric": [
        "Total Students",
        "Unique Job Professions",
        "Average Intelligence Score",
        "Average Performance Score",
        "Highest Average Intelligence Score",
        "Lowest Average Intelligence Score"
    ],
    "Value": [
        len(df),
        df["Job_Profession"].nunique(),
        round(df["Average_Intelligence_Score"].mean(), 2),
        round(df["Average_Performance"].mean(), 2),
        round(df["Average_Intelligence_Score"].max(), 2),
        round(df["Average_Intelligence_Score"].min(), 2)
    ]
})

student_statistics
```

```python
# Save statistics

output_path = "/content/drive/MyDrive/AI POWERED PERSONALISED LEARNING PATH GENERATOR/required datasets/student_statistics.csv"

student_statistics.to_csv(output_path, index=False)

print("Student Statistics saved successfully!")
```
