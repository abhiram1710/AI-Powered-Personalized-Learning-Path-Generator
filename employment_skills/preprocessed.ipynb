```python
courses = courses.drop(columns=["course_url"])
```

```python
print(courses.columns.tolist())
```

```python
reviews.columns.tolist()
```

```python
reviews.drop(columns=["reviewers", "date_reviews"], inplace=True)
```

```python
reviews.columns.tolist()
```

```python
merged = pd.merge(
    courses,
    reviews,
    on="course_id",
    how="inner"
)
merged
```

```python
merged.isnull().sum()
```

```python
merged.dropna(subset=["reviews"], inplace=True)
```

```python
merged.isnull().sum()
```

```python
merged.duplicated().sum()
```

```python
merged.drop_duplicates(inplace=True)
```

```python
merged.duplicated().sum()
```

```python
merged.info()
```

```python
#Remove Empty Reviews
merged = merged[
    merged["reviews"].str.len() > 15
]
```

```python
text_columns = ["name", "institution", "reviews"]

for col in text_columns:
    merged[col] = (
        merged[col]
        .astype(str)
        .str.lower()
        .str.strip()
    )
```

```python
text_columns
```

```python
#Remove Extra Spaces
for col in ["name", "institution", "reviews"]:
    merged[col] = merged[col].str.replace(r"\s+", " ", regex=True)
```

```python
merged.drop_duplicates(
    subset=["course_id", "reviews"],
    inplace=True
)
```
