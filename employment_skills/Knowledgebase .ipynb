```python
courses = pd.read_csv("processed_courses.csv")
```

```python
course_kb = courses.groupby(
    ["course_id", "name", "institution"]
).agg(
    average_rating=("rating", "mean"),
    review_count=("rating", "count"),
    reviews=("reviews", lambda x: " ".join(x.astype(str)))
).reset_index()
```

```python
course_kb.head()
```

```python
course_kb["reviews"] = course_kb["reviews"].str[:5000]
```

```python
course_kb["knowledge"] = (
    "Course Name: " + course_kb["name"] +
    ". Institution: " + course_kb["institution"] +
    ". Average Rating: " + course_kb["average_rating"].round(2).astype(str) +
    ". Number of Reviews: " + course_kb["review_count"].astype(str) +
    ". Reviews: " + course_kb["reviews"]
)
```

```python
course_kb[["knowledge"]].head()
```

```python
course_kb = course_kb[
    [
        "course_id",
        "name",
        "institution",
        "average_rating",
        "review_count",
        "knowledge"
    ]
]
```

```python
course_kb.to_csv("course_knowledge_base.csv", index=False)
```
